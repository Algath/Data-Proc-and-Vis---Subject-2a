# Initial Failure Rate by Year

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/vehicles-initial-failures-by-defect-category.csv")

failure_df = df[
    (df["Period"] == "Total")
    & (df["Defect category"] == "Overall Initial Failure Rate")
].copy()

failure_df["Overall % of tests"] = (
    failure_df["Overall % of tests"]
    .astype(str)
    .str.replace("%", "")
    .astype(float)
)

failure_df["year_start"] = failure_df["Financial year"].str.extract(r"^(\d{4})").astype(int)
failure_df = failure_df.sort_values("year_start")

plt.figure(figsize=(12, 5))
plt.bar(
    failure_df["Financial year"],
    failure_df["Overall % of tests"],
    color="#4C78A8",
)
plt.title("Total Initial Failure Rate by Financial Year")
plt.xlabel("Financial year")
plt.ylabel("Failure rate (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

print(failure_df[["Financial year", "Overall % of tests"]])
```
