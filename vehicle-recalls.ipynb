# Vehicle Recalls by Year

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/vehicle-recalls.csv")

year_counts = (
    df.dropna(subset=["recall_year"])
      .assign(recall_year=lambda frame: frame["recall_year"].astype(int))
      .groupby("recall_year")
      .size()
      .sort_index()
)

plt.figure(figsize=(12, 5))
plt.bar(year_counts.index.astype(str), year_counts.values, color="#4C78A8")
plt.title("Vehicle recall cases by year")
plt.xlabel("Year")
plt.ylabel("Number of cases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(year_counts.to_string())
```
